```
## demonstrate how to create simple and multi- line strings and whether a
## string can change after creation. Also show the usage of built in functions
## len(), min(), and max() on string.

msg1 = 'Hoopla'
print(msg1)
# strings with special characters
msg2 = 'He said, \'Let Us Python\'.'
file1 = 'C:\\temp\\newfile'
file2 = r'C:\temp\newfile'
print(msg2)
print(file1)
print(file2)

## multiline strings
## whitespace at beginning of second line becomes part of string
msg3 = 'What is this life if full of care ... \
        We have no time to stand and stare'

## enter at the end of frist line becomes part of string
msg4 = """What is this life if full of care...
          We have no time to stand and stare"""

## strings are concatenated properly.()necessary
msg5 = ('What is this life if full care...'
        'We have no time to stand and stare')

print(msg3)
print(msg4)
print(msg5)
```

```
import math

x = 3.7
y = math.trunc(x)  # truncates to 3
print(y)  # Output: 3

z = -2.9
print(math.trunc(z))
```

```
import math

x = 3.2
y = math.ceil(x)  # rounds up to 4
print(y)  # Output: 4

z = -2.7
print(math.ceil(z))
```

```
import math

x = 3.7
y = math.floor(x)  # rounds down to 3
print(y)  # Output: 3

z = -2.3
print(math.floor(z))  # rounds down to -3
```

```
num = int(input('Enter an integer: '))
i = 2
while i <= num - 1 :
    if num % i == 0 :
        print(num, 'is not a prime number')
        break
    i += 1
else :
    print(num, 'is a prime number')
```
