```
import random

# Step 1: Generate a list of 20 random integers between 1 and 100
numbers = [random.randint(1, 100) for _ in range(20)]
print("Generated List:", numbers)

# Step 2: Take input from the user
target = int(input("Enter a number to search for: "))

# Step 3: Find all positions (1-based index for user friendliness)
positions = [i + 1 for i, value in enumerate(numbers) if value == target]

# Step 4: Report results
if positions:
    print(f"The number {target} occurs at positions: {positions}")
else:
    print(f"The number {target} does not occur in the list.")
```

```
import random

# generate random list
lst = [random.randint(1, 50) for _ in range(20)]
print("List:", lst)

# input number
n = int(input("Enter a number: "))

# find positions
positions = []
for i in range(len(lst)):
    if lst[i] == n:
       positions.append(i)
# 0-based index

# output
if positions:
    print("Number found at positions:", positions)
else:
    print("Number not found in the list.")
```

```
# Create a list of 5 names
names = ['Anil', 'Amol', 'Aditya', 'Avi', 'Alka']
print(names)
# insert a name 'Anuj' before 'Aditya'
names.insert(2, 'Anuj')
print(names)
# append a name 'Zulu'
names.append('Zulu')
print(names)
# delete 'Avi' from the list
names.remove('Avi')
print(names)
# replace 'Anil' with 'AnilKumar'
i=names.index('Anil')
names[i] = 'AnilKumar'
print(names)
# sort all the names in the list
names.sort( )
print(names)
# print reversed sorted list
names.reverse()
print(names)
```
